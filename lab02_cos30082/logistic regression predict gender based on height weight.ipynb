```python
import numpy as np
import pandas as pd
from sklearn import linear_model
# Load data
file_name = "heights_weights.csv"
df = pd.read_csv(file_name)
df.head()
```

```python
df.shape
```

```python
# Plotting our data
import seaborn as sns; sns.set()
import matplotlib.pyplot as plt

ax = sns.scatterplot(x="Height", y="Weight", hue="Male", data=df)
```

```python
# Extract the columns we'll use for our data

x = df.iloc[:,0:2].values 
y = df.iloc[:,2].values
```

```python
# Split data into our test and training datasets
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(x, y, test_size=0.3, random_state=0)
```

```python
# Fit (train) the Logistic Regression classifier
clf = linear_model.LogisticRegression(C=1e40, solver='newton-cg')
fitted_model = clf.fit(X_train, Y_train)
```

```python
# Two persons' height and weight
new_data = [[74,240], [80, 50]]  # List of tuples (height, weight)

# Predict the outcome for both persons
predictions = clf.predict(new_data)

```

```python
# Mapping predictions to labels
for i, pred in enumerate(predictions):
    result = "Male" if pred else "Female"
    print(f"Person {i+1} is {result}")

```
